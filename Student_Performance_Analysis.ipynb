```python
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

```python
data = {
    "Name": ["Rahul", "Priya", "Arjun", "Sneha", "Kiran",
             "Anjali", "Rohit", "Divya", "Varun", "Meena"],

    "Gender": ["Male", "Female", "Male", "Female", "Male",
               "Female", "Male", "Female", "Male", "Female"],

    "Attendance": [90, 95, 75, 85, 60, 92, 70, 88, 65, 98],

    "Assignment": [85, 92, 70, 88, 65, 95, 72, 90, 60, 96],

    "Internal": [80, 90, 65, 82, 55, 94, 68, 87, 58, 97],

    "Final": [88, 94, 68, 86, 58, 96, 70, 91, 62, 98]
}

df = pd.DataFrame(data)

print(df)
```

```python

import matplotlib.pyplot as plt
import numpy as np

# Student names and final marks
names = df["Name"]
marks = df["Final"]

# Assign colors based on marks
colors = ["green" if mark == marks.max() else "skyblue"
          for mark in marks]

# Create chart
plt.figure(figsize=(12, 6))

bars = plt.bar(
    names,
    marks,
    color=colors,
    edgecolor="black",
    width=0.6
)

# Display marks above bars
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height + 1,
        f"{height:.0f}",
        ha="center",
        fontsize=10,
        fontweight="bold"
    )

# Average marks reference line
average = marks.mean()
plt.axhline(
    average,
    color="red",
    linestyle="--",
    label=f"Average: {average:.1f}"
)

# Titles and formatting
plt.title("Student Final Examination Performance",
          fontsize=16, fontweight="bold")

plt.xlabel("Student Names", fontsize=12)
plt.ylabel("Final Marks", fontsize=12)

plt.ylim(0, 110)
plt.xticks(rotation=30)
plt.grid(axis="y", linestyle=":", alpha=0.5)

plt.legend()
plt.tight_layout()

# Save chart
plt.savefig("professional_student_performance.png",
            dpi=300, bbox_inches="tight")

plt.show()
```

```python

import seaborn as sns
import matplotlib.pyplot as plt

# Select numerical columns
correlation_matrix = df[
    ["Attendance", "Assignment", "Internal", "Final"]
].corr()

# Create heatmap
plt.figure(figsize=(9, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    vmin=-1,
    vmax=1
)

plt.title("Student Performance Correlation Heatmap",
          fontsize=15, fontweight="bold")

plt.tight_layout()
plt.savefig("student_correlation_heatmap.png",
            dpi=300, bbox_inches="tight")
plt.show()
```

```python

import matplotlib.pyplot as plt
import numpy as np

fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# 1. Final marks
axes[0, 0].bar(df["Name"], df["Final"], color="steelblue")
axes[0, 0].set_title("Final Examination Marks")
axes[0, 0].set_ylabel("Marks")

# 2. Attendance
axes[0, 1].bar(df["Name"], df["Attendance"], color="seagreen")
axes[0, 1].set_title("Student Attendance")
axes[0, 1].set_ylabel("Attendance (%)")

# 3. Assignment marks
axes[1, 0].bar(df["Name"], df["Assignment"], color="orange")
axes[1, 0].set_title("Assignment Marks")
axes[1, 0].set_ylabel("Marks")

# 4. Internal marks
axes[1, 1].bar(df["Name"], df["Internal"], color="mediumpurple")
axes[1, 1].set_title("Internal Examination Marks")
axes[1, 1].set_ylabel("Marks")

# Format all charts
for ax in axes.flat:
    ax.tick_params(axis="x", rotation=45)
    ax.grid(axis="y", linestyle="--", alpha=0.3)

fig.suptitle(
    "STUDENT PERFORMANCE ANALYSIS DASHBOARD",
    fontsize=18,
    fontweight="bold"
)

plt.tight_layout()
plt.savefig("student_performance_dashboard.png",
            dpi=300, bbox_inches="tight")
plt.show()
```

```python

def categorize_performance(marks):
    if marks >= 85:
        return "Excellent"
    elif marks >= 70:
        return "Good"
    else:
        return "Needs Improvement"

df["Performance"] = df["Final"].apply(categorize_performance)

print(df[["Name", "Final", "Performance"]])
```

```python

import matplotlib.pyplot as plt

performance_counts = df["Performance"].value_counts()

plt.figure(figsize=(8, 5))

plt.bar(
    performance_counts.index,
    performance_counts.values,
    color=["seagreen", "skyblue", "salmon"]
)

plt.title("Student Performance Categories")
plt.xlabel("Performance Category")
plt.ylabel("Number of Students")

for i, value in enumerate(performance_counts.values):
    plt.text(i, value + 0.1, str(value), ha="center")

plt.tight_layout()
plt.savefig("performance_categories.png", dpi=300)
plt.show()
```
